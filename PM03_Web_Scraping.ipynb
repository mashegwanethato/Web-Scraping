```python
#importing beautiful soup, pandas and requests
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
#Collect URL
url = "https://books.toscrape.com/"
response = requests.get(url,timeout=20)

print(response.status_code)
```

```python
# Making the special character to be recognized
response.encoding = "utf-8-sig"

#Declaring object
Soup = BeautifulSoup(response.text,"html.parser")

#print results
print(Soup.title.text)
```

```python
#Declaring variable and retrieving book
books = Soup.select("article.product_pod")
#Getting results of the number od pages
print("Number of Books Found:", len(books))
```

```python
#Creating a for loop to get the book titles
for book in books:
  title = book.h3.a["title"]
  print(title)
```

```python
#Creating a for loop to get the price
for book in books:
  price = book.select_one(".price_color").get_text(strip=True) #.get_text(strip=True) - removes unnecessary space
  print(price)
```

```python
data = []

url = "https://books.toscrape.com/"

while url:

  response = requests.get(url, timeout=20)
  response.encoding = "utf-8-sig"

  Soup = BeautifulSoup(response.text, "html.parser")

  Books = Soup.select("article.product_pod")

  for book in Books:

    title = book.h3.a["title"]

    price = book.select_one(
        ".price_color"
    ).get_text(strip=True)

    price = float(price.replace("£", ""))

    rating = book.select_one(
        ".star-rating"
    )["class"][1]

    availability = book.select_one(
        ".availability"
    ).get_text(strip=True)

    data.append ({
        "Book Title": title,
        "Price": price,
        "Rating": rating,
        "Availability": availability
    })

  next_button = Soup.select_one("li.next a")

  if next_button:
      url = requests.compat.urljoin(url, next_button["href"])
  else:
      url = None

df = pd.DataFrame(data)

print("Total books scraped:", len(df))

display(df)
```

```python
#converting to a csv File
df.to_csv("Scraped_books.csv",
      index=False,
      encoding = "utf-8-sig"
  )

print("csv file saved successfully!")
```

```python
#Converts price into floating point
df["Price"] = df["Price"].str.replace("£", "", regex=False).astype(float)
display(df.head())
```

```python
#converting to a Excel File
df.to_excel("Scrapped_Books.xlsx",
      index=False,
)
print("Excel file saved successfully!")
```
